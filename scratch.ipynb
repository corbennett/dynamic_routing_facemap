```python
import dr_datacube
import numpy as np
import polars as pl

from matplotlib import pyplot as plt
```

```python
session_id = '626791_2022-08-15'
facemap_lf = dr_datacube.get_lf('processing/behavior/facemap_side_camera', session_id=session_id)
```

```python
trials_lf = dr_datacube.get_lf('trials')
```

```python
lick_trials = trials_lf.filter(pl.col('session_id') == session_id, (pl.col('is_instruction') == False), pl.col('is_response') == True).collect()
no_lick_trials = trials_lf.filter(pl.col('session_id') == session_id, (pl.col('is_instruction') == False), pl.col('is_response') == False).collect()
```

```python
lick_trials.filter(pl.col('is_vis_stim'))
```

```python
from scipy.interpolate import interp1d

window_to_capture = [-1, 1]
features_to_use = 200
sampling_rate = 60
n_samples = int((window_to_capture[1] - window_to_capture[0]) * sampling_rate)

condition_svds = {'lick': [], 'no_lick': []}
for condition_name, condition_trials in (('lick', lick_trials), ('no_lick', no_lick_trials)):

    trial_starts = condition_trials["stim_start_time"].to_numpy() + window_to_capture[0]
    target_times = (
        trial_starts[:, None] +
        np.arange(n_samples)[None, :] / sampling_rate
    )

    # One datacube read instead of one per trial
    facemap = (
        facemap_lf
        .filter(
            pl.col("timestamps").is_between(
                trial_starts.min(),
                trial_starts.max() + window_to_capture[1] - window_to_capture[0],
                closed="left",
            )
        )
        .select("timestamps", "data")
        .collect()
    )

    timestamps = facemap["timestamps"].to_numpy()
    data = np.asarray(facemap["data"].to_list())[:, :features_to_use]

    # Interpolate every trial and feature at once
    interpolator = interp1d(
        timestamps,
        data,
        axis=0,
        bounds_error=False,
        fill_value=(data[0], data[-1]),
    )

    # Shape: trials × features × samples
    condition_svds[condition_name]= interpolator(target_times).transpose(0, 2, 1)
```

```python
#suppress warnings
import warnings
warnings.filterwarnings('ignore')
```

```python
from sklearn.svm import LinearSVC
from sklearn.model_selection import cross_val_score
y = np.concatenate([np.ones(condition_svds['lick'].shape[0]), np.zeros(condition_svds['no_lick'].shape[0])])

windowlength = 1
window_scores = []
for windowstart in range(condition_svds['lick'].shape[2] - windowlength + 1):
    X = np.concatenate([condition_svds['lick'][:, :, windowstart:windowstart+windowlength], 
                        condition_svds['no_lick'][:, :, windowstart:windowstart+windowlength]], axis=0) 
    X_window = X.reshape(X.shape[0], -1)
    clf = LinearSVC()
    scores = cross_val_score(clf, X_window, y, cv=5)
    window_scores.append(scores.mean())
```

```python
time = ((np.arange(condition_svds['lick'].shape[2] - windowlength + 1) + np.floor(windowlength/2)) - 60)*(1/60)
plt.plot(time, window_scores)
plt.xlim(-0.4, 0.4)
```

```python
facemap = (
    facemap_lf
    .filter(
        pl.col("timestamps").is_between(
            1000,
            1100,
            closed="left",
        )
    )
    .select("timestamps", "data")
    .collect()
)
```

```python
comp_variances = np.var(facemap['data'].to_numpy(), axis=0)

variance_explained = (comp_variances / np.sum(comp_variances)) * 100
cumulative_variance = np.cumsum(variance_explained)

plt.plot(cumulative_variance)
plt.xlabel('Component')
plt.ylabel('Cumulative Variance Explained (%)')
```
